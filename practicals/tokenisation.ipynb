# Tokenisation

In this short practical we look at a few examples of tokenisation. You can try things like editing the strings to see how different kinds of strings are tokenised. You could try out long words, different capitalisation, including symbols, emoji even...

First let's look at `tiktoken` from OpenAI (See, https://github.com/openai/tiktoken).

You can install `tiktoken` with conda or pip. If you use conda, and `tiktoken` is not installed, you can uncomment the following line to install `tiktoken`. (To uncomment the line, remove the `#` symbol at the start of the line).

```python
#!yes|conda install tiktoken
```

```python
import tiktoken

enc = tiktoken.encoding_for_model("gpt-4o") # use this form to select by model name, or
# enc = tiktoken.get_encoding("o200k_base") # use this form to select by enconding name
```

In this step, we've selecting our **encoding**. This selects the vocabulary for the tokenisation. You should use the same encoding that was used when the model was trained (which is why it can be helpful to use the form above with the model name).

```python
text = "Welcome to EPCC at the University of Edinburgh"

tokens = enc.encode(text)

print(tokens)
```

```python
for token in tokens:
    print(f"{token}: '{enc.decode([token])}'")
```

Notice how most tokens start with a space, and how less common "words" like EPCC are split into multiple tokens.

```python
text = "your prompt is broken up into tokens"
tokens = enc.encode(text)
for token in tokens:
    print(f"{token}: '{enc.decode([token])}'")
```

```python
text = "Your prompt is broken up into tokens"
tokens = enc.encode(text)
for token in tokens:
    print(f"{token}: '{enc.decode([token])}'")
```

How many different tokens do you think there are in the vocabulary? (A clue is in the name of the encoding...)

We can see exactly how many there are as follows:

```python
print(enc.n_vocab)
```

You can experiment with encoding and decoding different strings. You will notice that some rare symbols are represented by multiple tokens. (If you dig into this in more detail, you can see that some token identifiers relate to the bytes that are used to represent the symbols in the Unicode encoding. This detail is not directly relevant but the key points are that (i) every unicode string can be encoded, and (ii) rare characters can be represented by multiple tokens.

```python
enc.decode([4103,99,117,52622,121,2524,15794,15148])
```

```python
enc.decode(enc.encode("🦹🏽‍♀️"))
```
